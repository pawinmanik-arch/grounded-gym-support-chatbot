```python
knowledge_base = """
GYM SUPPORT KNOWLEDGE BASE
Gym Name: FitCare Fitness Centre

AREA 1: MEMBERSHIP PLANS

Q: What membership plans do you offer?
A: We offer monthly, quarterly, and yearly membership plans.

Q: How much does a monthly membership cost?
A: The monthly membership costs Rs. 1500.

Q: Do you offer a yearly membership?
A: Yes, the yearly membership costs Rs. 15000 and lasts 12 months.

AREA 2: TIMINGS AND ACCESS

Q: What time does the gym open?
A: The gym opens at 6:00 AM.

Q: What time does the gym close?
A: The gym closes at 10:00 PM.

Q: Is the gym open on Sundays?
A: Yes, the gym is open on Sundays from 7:00 AM to 12:00 PM.

AREA 3: TRAINERS AND CLASSES

Q: Is a personal trainer available?
A: Yes, personal training is available by appointment.

Q: How much does personal training cost?
A: A personal training session costs Rs. 500 for 60 minutes.

Q: Do you conduct group fitness classes?
A: Yes, group fitness classes are held Monday to Saturday at 7:00 AM.

AREA 4: FACILITIES AND RULES

Q: What equipment is available at the gym?
A: The gym has treadmills, exercise bikes, dumbbells, barbells, and weight machines.

Q: Do you have changing rooms and showers?
A: Yes, changing rooms and showers are available.

Q: Is there a minimum age for membership?
A: Members must be at least 16 years old.

AREA 5: PAYMENTS AND MEMBERSHIP CHANGES

Q: What payment methods do you accept?
A: We accept cash, UPI, and debit or credit cards.

Q: Can I cancel my membership?
A: Membership cancellation requires 7 days of advance notice.

Q: Can I pause or freeze my membership?
A: Memberships can be paused once for up to 14 days per year.
"""
print(knowledge_base)
```

```python
system_instruction = f"""
You are FitCare Support, a customer-support chatbot
for FitCare Fitness Centre.

Answer questions ONLY using the knowledge base below.

Rules:
1. Never invent facts or policies.
2. If the answer is missing from the knowledge base,
   reply with this exact sentence:
   "I'm sorry, I don't have that information in my
   knowledge base. Please contact the gym at
   +91-9489301929 or pawinmanik@gmail.com."
3. For unrelated questions, use the same exact
   fallback sentence.
4. Use previous messages when relevant.
5. Be polite, clear, and concise.

KNOWLEDGE BASE:
{knowledge_base}
"""

print("System instruction created successfully!")
```

```python
from google import genai
from google.genai import types
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)

print("Gemini client connected successfully!")
```

```python
chat = client.chats.create(
    model="gemini-2.5-flash",
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
        temperature=0.2
    )
)

print("FitCare Support chatbot is ready!")
```

```python
print("Welcome to FitCare Support!")
print("Ask me about memberships, timings, trainers,")
print("facilities, or gym policies.")
print("Type 'quit' or 'exit' to stop.\n")

while True:
    user_input = input("You: ").strip()

    if user_input.lower() in ["quit", "exit"]:
        print("FitCare Support: Thank you for contacting us!")
        break

    if not user_input:
        continue

    try:
        response = chat.send_message(user_input)
        print("FitCare Support:", response.text, "\n")

    except Exception as error:
        print("An error occurred:", error)
```
